```python
# Functional Requirement Analysis
# Compute mean scores o determine highest/lowest rated requirements
fr_cols = [
    'fr_auth',
    'fr_role_mgmt',
    'fr_student_mgmt',
    'fr_hte_mgmt',
    'fr_internship_mgmt',
    'fr_internship_status_tracking',
    'fr_attendance_mgmt',
    'fr_evaluation_mgmt',
    'fr_document_mgmt',
    'fr_reporting',
    'fr_audit_logging'
]

fr_means = df[fr_cols].mean().sort_values(ascending=False)
print(fr_means)
```

```python
# Non-functional Requirement Analysis
nfr_cols = [
    'nfr_security',
    'nfr_reliability',
    'nfr_usability',
    'nfr_compliance'
]

nfr_means = df[nfr_cols].mean().sort_values(ascending=False)
print(nfr_means)
```

```python
# Overall Requirement Validity
# Compute overall average score or the Overall weighted mean
all_req_cols = fr_cols + nfr_cols
overall_mean = df[all_req_cols].mean().mean()
print(overall_mean)
```

```python
# Approval Analysis
df['approval'].value_counts()
approval_pct = df['approval'].value_counts(normalize=True) * 100
print(approval_pct)
```

```python
# Combined approval as a key meric
accepted = df['approval'].isin(['approved', 'approved_revision']).sum()
total = len(df)

print((accepted / total) * 100)
```

```python
# Group Analysis (compare by role)
# Ex: Do faculty rate differently from coordinators?
df.groupby('role_group')[fr_cols + nfr_cols].mean()
```
