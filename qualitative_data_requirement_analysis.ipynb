# Qualitative Data Analysis

> Since the open-ended responses are relatively small, the approach is manual thematic analysis assisted by Python. The goal is to transform open-ended responses into: themes, categories and actionable requirement revisions.

```python
import pandas as pd

df = pd.read_csv("datasets/qualitative_data.csv")
df.head()
```

```python
# Keep Only Meaningful Responses
qual = df[
    df["response_quality"] != "insufficient_context"
].copy()

print("Number of substantive responses:", len(qual))
```

```python
# Merge the Four Open-ended Questions
# Instead of coding four columns separately, treat each respondent's comments as one transcript.
text_cols = [
    "unnecessary_requirements",
    "improvement_suggestions",
    "missing_features",
    "comments"
]

qual["full_response"] = (
    qual[text_cols]
    .fillna("")
    .agg(" ".join, axis=1)
)

qual[["respondent_id","full_response"]]
```

```python
# Display One Response at a Time
for _, row in qual.iterrows():
    print("="*80)
    print(row["respondent_id"])
    print()
    print(row["full_response"])
    print()
```

```python
# Open Coding Table

codes = {

    "R01":[
        "HTE Role Simplification",
        "Audit Trail Support",
        "Attendance Verification",
        "External Evaluation Process",
        "Mentor Coordination"
    ],

    "R02":[
        "Notification Support",
        "Automated Notifications",
        "Reminder System"
    ],

    "R03":[
        "Role Definition",
        "Status Tracking Clarification",
        "Notification Support",
        "Dashboard",
        "Calendar Management",
        "Messaging",
        "Mobile Accessibility",
        "Data Backup"
    ],

    "R04":[
        "HTE Access Control",
        "Role Definition",
        "Document Management"
    ],

    "R05":[
        "No Actionable Feedback"
    ],

    "R06":[
        "Minimalist User Interface"
    ],

    "R07":[
        "Student Information Integration",
        "Authentication Clarification",
        "Role Definition",
        "Attendance Workflow",
        "Security Controls",
        "Reliability Requirements",
        "Usability Clarification",
        "Compliance Clarification",
        "Application Workflow",
        "Notification Support",
        "Internship Opportunity Management",
        "Supervisor Assignment",
        "Document Approval Workflow",
        "Certificate Generation",
        "Dashboard"
    ],

    "R09":[
        "Requirement Consolidation",
        "Document Version Control",
        "Digital Signature",
        "Customizable Reporting",
        "Notification Support",
        "System Integration",
        "Mobile Accessibility",
        "Analytics",
        "Scalability",
        "Audit Logging"
    ],

    "R10":[
        "System Security",
        "Data Protection"
    ],

    "R11":[
        "Activity Notifications",
        "Role-Based Access"
    ]
}
```

```python
# Convert to DF to create the open code table
coding_table = []

for respondent, code_list in codes.items():

    for code in code_list:

        coding_table.append({
            "Respondent": respondent,
            "Open Code": code
        })

coding_df = pd.DataFrame(coding_table)

coding_df
```

```python
# Count Code Frequencies
code_frequency = (
    coding_df
    .value_counts("Open Code")
    .reset_index(name="Frequency")
    .sort_values("Frequency", ascending=False)
)

code_frequency
```
